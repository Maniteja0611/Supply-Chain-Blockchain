## 1. Problem Statement

Traditional supply chain management systems often maintain raw-material and product records in centralized databases. These records may be modified without proper authorization, making it difficult to establish reliable product histories.

When a raw material is found to be contaminated, manufacturers may struggle to identify all finished products manufactured using that particular batch and trace their distribution to retailers.

This project proposes a Python-based blockchain system that maintains tamper-evident supply chain records. It tracks raw materials from suppliers through manufacturing and distribution.

When contamination is reported, the system identifies the affected raw-material batch, traces all associated finished products, identifies their distribution destinations, and generates a product recall report.

## 2. Input Specification (I/P)

1. Student roll number.
2. Supplier details and raw-material batch ID.
3. Raw-material details, including product type and quantity.
4. Manufacturing details, including raw-material batch IDs and finished-product IDs.
5. Distribution details, including distributor and retailer information.
6. Contamination report containing the affected batch ID and reason.
7. Requester identity and authorization details.

## 3. Output Specification (O/P)

1. Unique SHA-256 hashes for supply-chain transactions.
2. Blockchain records containing transaction details and previous block hashes.
3. Raw-material and finished-product traceability records.
4. Identification of all finished products linked to a contaminated batch.
5. Recall report containing affected products, distributors, and retailers.
6. Blockchain integrity verification and tampering detection status.
7. Transaction execution receipts.

## 4. Target Challenges

- Data Tampering: Detect unauthorized modification of supply-chain records.
- Traceability: Establish relationships between raw-material batches and finished products.
- Unauthorized Access: Restrict sensitive operations to authorized entities.
- Product Recall: Identify all affected products and their distribution destinations.
- Data Integrity: Maintain cryptographic hash linkage between blockchain blocks.
- Accountability: Maintain a verifiable history of supply-chain transactions.

```python
class UnauthorizedSupplyChainAccess(Exception):

    def __init__(self, message):
        self.message = message
        super().__init__(self.message)
```

```python
def verify_permission(caller_id, allowed_roles):

    users = world_state["users"]

    # Check whether user exists
    if caller_id not in users:
        raise UnauthorizedSupplyChainAccess(
            "User does not exist in the supply chain ledger"
        )

    # Retrieve user role
    user_role = users[caller_id]["role"]

    # Verify permission
    if user_role not in allowed_roles:
        raise UnauthorizedSupplyChainAccess(
            f"Access Denied: {user_role} is not authorized"
        )

    return True
```

```python
def register_raw_material(caller_id, batch_id, material_name, quantity, unit):

    # Verify supplier permission
    verify_permission(caller_id, ["SUPPLIER"])

    # Check duplicate batch ID
    if batch_id in world_state["raw_materials"]:
        raise ValueError("Batch ID already exists")

    # Validate quantity
    if quantity <= 0:
        raise ValueError("Quantity must be greater than zero")

    # Create raw material record
    raw_material = {
        "batch_id": batch_id,
        "supplier_id": caller_id,
        "material_name": material_name,
        "quantity": quantity,
        "unit": unit,
        "status": "APPROVED",
        "timestamp": datetime.now().isoformat()
    }

    # Generate transaction hash
    transaction_hash = generate_hash(raw_material)

    # Update global ledger
    world_state["raw_materials"][batch_id] = raw_material

    # Create execution receipt
    receipt = {
        "transaction": "RAW_MATERIAL_REGISTERED",
        "batch_id": batch_id,
        "transaction_hash": transaction_hash,
        "status": "SUCCESS"
    }

    world_state["transactions"].append(receipt)

    return receipt
```

```python
def manufacture_product(caller_id, product_id, raw_material_batches, quantity):

    # Verify manufacturer permission
    verify_permission(caller_id, ["MANUFACTURER"])

    # Check duplicate product ID
    if product_id in world_state["finished_products"]:
        raise ValueError("Product ID already exists")

    # Validate quantity
    if quantity <= 0:
        raise ValueError("Quantity must be greater than zero")

    # Verify raw material batches
    for batch_id in raw_material_batches:

        if batch_id not in world_state["raw_materials"]:
            raise ValueError(f"Batch {batch_id} does not exist")

        if world_state["raw_materials"][batch_id]["status"] != "APPROVED":
            raise ValueError(f"Batch {batch_id} is not approved")

    # Create finished product record
    product = {
        "product_id": product_id,
        "raw_material_batches": raw_material_batches,
        "manufacturer_id": caller_id,
        "quantity": quantity,
        "status": "MANUFACTURED",
        "timestamp": datetime.now().isoformat()
    }

    # Generate transaction hash
    transaction_hash = generate_hash(product)

    # Update global ledger
    world_state["finished_products"][product_id] = product

    # Generate execution receipt
    receipt = {
        "transaction": "PRODUCT_MANUFACTURED",
        "product_id": product_id,
        "transaction_hash": transaction_hash,
        "status": "SUCCESS"
    }

    world_state["transactions"].append(receipt)

    return receipt
```

```python
def report_contamination(caller_id, batch_id, reason):

    # Verify quality authority permission
    verify_permission(caller_id, ["QUALITY_AUTHORITY"])

    # Check whether batch exists
    if batch_id not in world_state["raw_materials"]:
        raise ValueError("Raw material batch does not exist")

    # Update batch status
    world_state["raw_materials"][batch_id]["status"] = "CONTAMINATED"

    # Create contamination record
    contamination = {
        "batch_id": batch_id,
        "reported_by": caller_id,
        "reason": reason,
        "status": "CONTAMINATED",
        "timestamp": datetime.now().isoformat()
    }

    transaction_hash = generate_hash(contamination)

    world_state["recalls"][batch_id] = contamination

    receipt = {
        "transaction": "CONTAMINATION_REPORTED",
        "batch_id": batch_id,
        "transaction_hash": transaction_hash,
        "status": "SUCCESS"
    }

    world_state["transactions"].append(receipt)

    return receipt
```

```python
def generate_recall_report(batch_id):

    # Verify contaminated batch
    if batch_id not in world_state["recalls"]:
        raise ValueError("No contamination report found")

    affected_products = []

    # Find products manufactured using contaminated batch
    for product_id, product in world_state["finished_products"].items():

        if batch_id in product["raw_material_batches"]:

            affected_products.append({
                "product_id": product_id,
                "quantity": product["quantity"],
                "destinations": []
            })

    # Find distribution destinations
    for distribution in world_state["distributions"].values():

        for product in affected_products:

            if distribution["product_id"] == product["product_id"]:

                product["destinations"].append({
                    "distributor_id": distribution["distributor_id"],
                    "retailer_id": distribution["retailer_id"],
                    "quantity": distribution["quantity"]
                })

    recall_report = {
        "contaminated_batch": batch_id,
        "reason": world_state["recalls"][batch_id]["reason"],
        "affected_products": affected_products,
        "total_affected_products": len(affected_products),
        "recall_status": "RECALL_REQUIRED",
        "timestamp": datetime.now().isoformat()
    }

    return recall_report
```

```python
recall_report = generate_recall_report("RM101")

print("========== PRODUCT RECALL REPORT ==========")

print(json.dumps(recall_report, indent=4))
```

```python
blockchain = []

def create_block(index, payload, previous_hash):
    block_data = {
        "index": index,
        "timestamp": datetime.now().isoformat(),
        "payload": payload,
        "previous_hash": previous_hash
    }

    block_data["current_hash"] = generate_hash(block_data)

    return block_data
```

```python
genesis_block = create_block(
    0,
    {"message": "Genesis Block"},
    "0" * 64
)

blockchain.append(genesis_block)

print("Genesis Block Created Successfully")
print(blockchain[0])
```

```python
def add_block(payload):
    previous_block = blockchain[-1]

    new_block = create_block(
        len(blockchain),
        payload,
        previous_block["current_hash"]
    )

    blockchain.append(new_block)

    return new_block
```

```python
for transaction in world_state["transactions"]:
    add_block(transaction)

print("All transactions added to blockchain!")
print("Total blocks:", len(blockchain))
```

```python
def verify_blockchain():
    for i in range(len(blockchain)):
        block = blockchain[i]

        # Recalculate the current block's hash
        block_data = {
            "index": block["index"],
            "timestamp": block["timestamp"],
            "payload": block["payload"],
            "previous_hash": block["previous_hash"]
        }

        recalculated_hash = generate_hash(block_data)

        if block["current_hash"] != recalculated_hash:
            return False, f"Block {i} has been tampered with!"

        # Verify connection with the previous block
        if i > 0:
            if block["previous_hash"] != blockchain[i - 1]["current_hash"]:
                return False, f"Block {i} has an invalid previous hash!"

    return True, "Blockchain is valid!"
```
